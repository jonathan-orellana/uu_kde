```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/heart_failure_clinical_records_dataset.csv")
df.head()
```

1. Plot a kernel density estimate of `ejection_fraction`.

```python
sns.kdeplot(data=df, x="ejection_fraction")

plt.xlabel("Ejection Fraction")
plt.ylabel("Density")
plt.title("KDE of Ejection Fraction")

plt.show()
```

2. Use Pandas' `df.sample(frac=1.0, replace=True)` to resample the data 15 times with replacement, plotting the kernel density estimtes for each sample.

```python
plt.figure(figsize=(8, 5))

for i in range(15):
    sample = df.sample(frac=1.0, replace=True)
    sns.kdeplot(data=sample, x="ejection_fraction")

plt.xlabel("Ejection Fraction")
plt.ylabel("Density")
plt.title("KDE of 15 Resampled Datasets")

plt.show()
```

3. For what values of `ejection_fraction` is the original plot reliable? For which values is there noticeable variation in your plots of the resampled data?
